```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import os
dosya_yolu = 'localDB/rgi_sonuclari.txt'

if os.path.exists(dosya_yolu):
    df_results = pd.read_csv(dosya_yolu, sep='\t')

sutunlar = df_results.columns.tolist()
best_hit_col = 'Best_Hit_ARO' if 'Best_Hit_ARO' in sutunlar else 'Best_Hit_ARO'

secilen_sutunlar = [
    'ORF_ID', 'Best_Hit_ARO', 'Drug Class',
    'Resistance Mechanism', 'AMR Gene Family', 'Best_Identities'
]

df_sade = df_results[secilen_sutunlar]
display(df_sade.head(10))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.countplot(
    data=df_results,
    y='Drug Class',
    order=df_results['Drug Class'].value_counts().index[:10],
    ax=axes[0],
    palette='Blues_r'
)
axes[0].set_title('En Sık Tespit Edilen İlaç Sınıfları', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Gen Sayısı')
axes[0].set_ylabel('İlaç Sınıfı')

sns.countplot(
    data=df_results,
    y='Resistance Mechanism',
    order=df_results['Resistance Mechanism'].value_counts().index[:10],
    ax=axes[1],
    palette='Reds_r'
)
axes[1].set_title('En Sık Tespit Edilen Direnç Mekanizmaları', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Gen Sayısı')
axes[1].set_ylabel('Direnç Mekanizması')

plt.tight_layout()
plt.show()
```
